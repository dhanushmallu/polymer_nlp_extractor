````python
# 2. Enhanced Dependencies Installation & Database Setup
import subprocess
import sys
import importlib

def install_enhanced_requirements():
    """Install comprehensive requirements for production training."""
    print("📦 Installing Enhanced Production Requirements...")
    
    # Core requirements from requirements.txt
    requirements_path = os.path.join(PROJECT_ROOT, 'requirements.txt')
    if os.path.exists(requirements_path):
        print("⬇️ Installing base requirements...")
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-r', requirements_path])
        print("✅ Base requirements installed")
    
    # Additional enhanced training requirements
    enhanced_packages = [
        "datasets>=2.14.0",  # Enhanced dataset handling
        "evaluate>=0.4.0",   # Model evaluation metrics
        "seqeval>=1.2.2",    # NER evaluation metrics
        "scikit-learn>=1.3.0",  # Advanced metrics
        "wandb>=0.15.0",     # Training monitoring (optional)
        "neptune-client>=1.7.0",  # Alternative monitoring
        "transformers[torch]>=4.30.0",  # Enhanced transformers
        "torch>=2.0.0",      # Latest PyTorch
        "torchtext>=0.15.0", # Text processing utilities
        "psycopg2-binary>=2.9.0",  # PostgreSQL connector
        "neo4j>=5.8.0",      # Neo4j driver
        "lxml>=4.9.0",       # XML processing
        "pandas>=2.0.0",     # Data manipulation
        "numpy>=1.24.0",     # Numerical computing
        "matplotlib>=3.7.0", # Visualization
        "seaborn>=0.12.0",   # Statistical visualization
        "tqdm>=4.65.0",      # Progress bars
        "python-dotenv>=1.0.0",  # Environment management
    ]
    
    for package in enhanced_packages:
        try:
            print(f"📥 Installing {package}...")
            subprocess.check_call([
                sys.executable, '-m', 'pip', 'install', 
                package, '--upgrade', '--quiet'
            ])
        except subprocess.CalledProcessError as e:
            print(f"⚠️ Failed to install {package}: {e}")
    
    print("✅ Enhanced requirements installation completed")

def setup_database_connections():
    """Setup database connections for training data access."""
    print("\n🗄️ Setting up Database Connections...")
    
    # Import database clients
    try:
        sys.path.append(os.path.join(PROJECT_ROOT, 'polymer_extractor'))
        
        # Import database managers
        from storage.database_manager import DatabaseManager
        from storage.neo4j_client import Neo4jClient
        
        # Initialize database manager
        db_manager = DatabaseManager()
        print("✅ DatabaseManager initialized")
        
        # Initialize Neo4j client if enabled
        if PRODUCTION_CONFIG["database"]["neo4j_enabled"]:
            neo4j_client = Neo4jClient()
            print("✅ Neo4j client initialized")
        
        # Store in global scope for later use
        globals()['db_manager'] = db_manager
        if PRODUCTION_CONFIG["database"]["neo4j_enabled"]:
            globals()['neo4j_client'] = neo4j_client
        
        return True
    except Exception as e:
        print(f"⚠️ Database setup failed: {e}")
        print("📝 Training will proceed without database integration")
        return False

def validate_training_environment():
    """Validate that all required components are available."""
    print("\n🔍 Validating Training Environment...")
    
    required_modules = [
        ('transformers', '4.30.0'),
        ('torch', '2.0.0'),
        ('datasets', '2.14.0'),
        ('pandas', '2.0.0'),
        ('numpy', '1.24.0'),
        ('sklearn', '1.3.0')
    ]
    
    validation_results = {}
    
    for module_name, min_version in required_modules:
        try:
            module = importlib.import_module(module_name.replace('-', '_'))
            if hasattr(module, '__version__'):
                version = module.__version__
                validation_results[module_name] = {
                    'available': True,
                    'version': version
                }
                print(f"✅ {module_name}: {version}")
            else:
                validation_results[module_name] = {
                    'available': True,
                    'version': 'unknown'
                }
                print(f"✅ {module_name}: available (version unknown)")
        except ImportError:
            validation_results[module_name] = {
                'available': False,
                'version': None
            }
            print(f"❌ {module_name}: not available")
    
    # Check for GPU availability
    try:
        import torch
        if torch.cuda.is_available():
            gpu_count = torch.cuda.device_count()
            gpu_name = torch.cuda.get_device_name(0) if gpu_count > 0 else "Unknown"
            print(f"🚀 GPU available: {gpu_count} device(s) - {gpu_name}")
            validation_results['gpu'] = {
                'available': True,
                'count': gpu_count,
                'name': gpu_name
            }
        else:
            print("💻 CPU training mode (no GPU detected)")
            validation_results['gpu'] = {'available': False}
    except ImportError:
        print("❌ PyTorch not available")
        validation_results['gpu'] = {'available': False}
    
    return validation_results

# Execute installation and setup
print("🚀 Enhanced Dependencies & Database Setup")
print("=" * 50)

# Install requirements
install_enhanced_requirements()

# Setup database connections
db_setup_success = setup_database_connections()

# Validate environment
validation_results = validate_training_environment()

# Global configuration update
PRODUCTION_CONFIG['database_available'] = db_setup_success
PRODUCTION_CONFIG['validation_results'] = validation_results

print("\n📊 Setup Summary:")
print(f"✅ Enhanced packages: Installed")
print(f"✅ Database connectivity: {'Available' if db_setup_success else 'Limited'}")
print(f"✅ GPU acceleration: {'Available' if validation_results.get('gpu', {}).get('available') else 'CPU only'}")
print(f"✅ Training environment: {'Validated' if all(v.get('available', False) for k, v in validation_results.items() if k != 'gpu') else 'Partial'}")

if not db_setup_success:
    print("\n⚠️ Database integration limited - training will use local data only")
    print("💡 For full functionality, ensure PostgreSQL and Neo4j are accessible")

print("✅ Enhanced setup completed - ready for production training")
print("=" * 50)
````

````python
# 5. Production Model Training with Enhanced Validation (Document 13)
from transformers import TrainingArguments, Trainer, DataCollatorForTokenClassification
from transformers import EarlyStoppingCallback
import torch
from torch.utils.data import DataLoader
from sklearn.metrics import classification_report, f1_score
import wandb  # Optional: for training monitoring

class ProductionModelTrainer:
    """
    Production-grade model training implementing Document 13 specifications.
    
    Features:
    - Model-tokenizer compatibility enforcement
    - Knowledge graph-enhanced training (configurable)
    - Entity boundary preservation validation
    - Comprehensive error handling and monitoring
    - Production-ready quality assurance
    """
    
    def __init__(self, compatibility_results: Dict[str, Any], kg_enhancer=None):
        self.compatibility_results = compatibility_results
        self.kg_enhancer = kg_enhancer
        self.training_results = {}
        
        # Enhanced training configuration
        self.training_config = {
            "output_dir": "./models/enhanced_training",
            "learning_rate": 2e-5,
            "per_device_train_batch_size": 16,
            "per_device_eval_batch_size": 16,
            "num_train_epochs": 5,
            "weight_decay": 0.01,
            "logging_dir": "./logs",
            "logging_steps": 100,
            "evaluation_strategy": "steps",
            "eval_steps": 500,
            "save_strategy": "steps",
            "save_steps": 500,
            "load_best_model_at_end": True,
            "metric_for_best_model": "eval_f1",
            "greater_is_better": True,
            "report_to": None,  # Disable wandb by default
            "push_to_hub": False
        }
    
    def prepare_training_data(self, dataset_results: Dict[str, Any]) -> Dataset:
        """
        Prepare training data with Document 13 format compliance and validation.
        """
        print("📊 Preparing Enhanced Training Data...")
        
        # Load synthetic dataset
        synthetic_path = dataset_results["synthetic"]["dataset_path"]
        df = pd.read_csv(synthetic_path)
        
        # Convert to training format with BIO tagging
        training_examples = []
        
        for _, row in df.iterrows():
            sentence = str(row["sentence"])
            polymer = str(row["polymer"])
            property_name = str(row["property"])
            value = str(row["value"])
            unit = str(row["unit"])
            symbol = str(row["symbol"])
            
            # Create BIO tags for the sentence
            tokens = sentence.split()
            labels = ["O"] * len(tokens)
            
            # Simple entity tagging (improved version would use character positions)
            for i, token in enumerate(tokens):
                token_lower = token.lower().strip(".,!?")
                
                # Tag polymer entities
                if polymer.lower() in token_lower:
                    labels[i] = "B-POLYMER" if i == 0 or labels[i-1] != "I-POLYMER" else "I-POLYMER"
                
                # Tag property entities
                elif any(prop_word in token_lower for prop_word in property_name.lower().split()):
                    labels[i] = "B-PROPERTY" if i == 0 or labels[i-1] != "I-PROPERTY" else "I-PROPERTY"
                
                # Tag value entities (numerical values only)
                elif value and value in token:
                    labels[i] = "B-VALUE"
                
                # Tag unit entities
                elif unit and unit in token:
                    labels[i] = "B-UNIT"
                
                # Tag symbol entities
                elif symbol and symbol in token:
                    labels[i] = "B-SYMBOL"
            
            example = {
                "tokens": tokens,
                "labels": labels,
                "sentence": sentence
            }
            
            # Apply knowledge graph enhancement if available
            if self.kg_enhancer:
                example = self.kg_enhancer.enhance_training_example(example)
            
            training_examples.append(example)
        
        print(f"✅ Prepared {len(training_examples)} training examples")
        
        # Convert to Hugging Face Dataset format
        dataset = Dataset.from_list(training_examples)
        
        return dataset
    
    def tokenize_and_align_labels(self, examples, tokenizer):
        """
        Tokenize examples and align labels with subword tokens.
        Implements entity boundary preservation from Document 12.
        """
        tokenized_inputs = tokenizer(
            examples["tokens"],
            truncation=True,
            is_split_into_words=True,
            padding=True,
            max_length=512
        )
        
        labels = []
        for i, label in enumerate(examples["labels"]):
            word_ids = tokenized_inputs.word_ids(batch_index=i)
            previous_word_idx = None
            label_ids = []
            
            for word_idx in word_ids:
                if word_idx is None:
                    label_ids.append(-100)
                elif word_idx != previous_word_idx:
                    # First subword of a word gets the label
                    label_ids.append(self._get_label_id(label[word_idx]))
                else:
                    # Subsequent subwords get -100 (ignored in loss)
                    label_ids.append(-100)
                previous_word_idx = word_idx
            
            labels.append(label_ids)
        
        tokenized_inputs["labels"] = labels
        return tokenized_inputs
    
    def _get_label_id(self, label: str) -> int:
        """Convert label string to ID."""
        label_map = {
            "O": 0,
            "B-POLYMER": 1, "I-POLYMER": 2,
            "B-PROPERTY": 3, "I-PROPERTY": 4,
            "B-VALUE": 5, "I-VALUE": 6,
            "B-UNIT": 7, "I-UNIT": 8,
            "B-SYMBOL": 9, "I-SYMBOL": 10
        }
        return label_map.get(label, 0)
    
    def compute_metrics(self, eval_pred):
        """Compute comprehensive evaluation metrics."""
        predictions, labels = eval_pred
        predictions = np.argmax(predictions, axis=2)
        
        # Remove ignored index (special tokens)
        true_predictions = [
            [p for (p, l) in zip(prediction, label) if l != -100]
            for prediction, label in zip(predictions, labels)
        ]
        true_labels = [
            [l for (p, l) in zip(prediction, label) if l != -100]
            for prediction, label in zip(predictions, labels)
        ]
        
        # Flatten for sklearn metrics
        flat_true = [item for sublist in true_labels for item in sublist]
        flat_pred = [item for sublist in true_predictions for item in sublist]
        
        # Calculate F1 score
        f1 = f1_score(flat_true, flat_pred, average='weighted', zero_division=0)
        
        return {
            "f1": f1,
            "accuracy": (np.array(flat_true) == np.array(flat_pred)).mean()
        }
    
    def train_single_model(
        self, 
        model_name: str, 
        training_dataset: Dataset,
        validation_split: float = 0.2
    ) -> Dict[str, Any]:
        """
        Train a single model with comprehensive validation and monitoring.
        """
        print(f"🔥 Training {model_name}...")
        
        # Get model and tokenizer from compatibility results
        model_result = self.compatibility_results["individual_results"][model_name]
        
        if not model_result.get("vocab_compatible", False):
            print(f"❌ Skipping {model_name} - compatibility issues")
            return {"success": False, "error": "Compatibility issues"}
        
        model = model_result["model"]
        tokenizer = model_result["tokenizer"]
        
        # Tokenize dataset
        tokenized_dataset = training_dataset.map(
            lambda examples: self.tokenize_and_align_labels(examples, tokenizer),
            batched=True
        )
        
        # Split into train/validation
        train_size = int(len(tokenized_dataset) * (1 - validation_split))
        train_dataset = tokenized_dataset.select(range(train_size))
        eval_dataset = tokenized_dataset.select(range(train_size, len(tokenized_dataset)))
        
        # Data collator
        data_collator = DataCollatorForTokenClassification(tokenizer=tokenizer)
        
        # Training arguments with model-specific configuration
        model_training_config = model_result["training_config"]
        training_args = TrainingArguments(
            output_dir=f"./models/{model_name}_enhanced",
            learning_rate=model_training_config["learning_rate"],
            per_device_train_batch_size=model_training_config["batch_size"],
            per_device_eval_batch_size=model_training_config["batch_size"],
            num_train_epochs=model_training_config["epochs"],
            weight_decay=0.01,
            evaluation_strategy="steps",
            eval_steps=200,
            logging_steps=50,
            save_strategy="steps",
            save_steps=200,
            load_best_model_at_end=True,
            metric_for_best_model="eval_f1",
            greater_is_better=True,
            report_to=None
        )
        
        # Initialize trainer
        trainer = Trainer(
            model=model,
            args=training_args,
            train_dataset=train_dataset,
            eval_dataset=eval_dataset,
            tokenizer=tokenizer,
            data_collator=data_collator,
            compute_metrics=self.compute_metrics,
            callbacks=[EarlyStoppingCallback(early_stopping_patience=3)]
        )
        
        try:
            # Train model
            print(f"🚀 Starting training for {model_name}...")
            training_result = trainer.train()
            
            # Evaluate model
            eval_result = trainer.evaluate()
            
            # Save model and tokenizer
            model_save_path = f"./models/{model_name}_enhanced"
            trainer.save_model(model_save_path)
            tokenizer.save_pretrained(model_save_path)
            
            print(f"✅ {model_name} training completed - F1: {eval_result['eval_f1']:.4f}")
            
            return {
                "success": True,
                "model_name": model_name,
                "save_path": model_save_path,
                "training_result": training_result,
                "eval_result": eval_result,
                "f1_score": eval_result['eval_f1'],
                "accuracy": eval_result['eval_accuracy']
            }
            
        except Exception as e:
            print(f"❌ Training failed for {model_name}: {e}")
            return {
                "success": False,
                "model_name": model_name,
                "error": str(e)
            }
    
    def train_all_compatible_models(self, training_dataset: Dataset) -> Dict[str, Any]:
        """Train all compatible models in the ensemble."""
        print("🚀 Training All Compatible Models")
        print("=" * 50)
        
        training_results = {}
        successful_models = []
        
        for model_name in self.compatibility_results["individual_results"].keys():
            if self.compatibility_results["individual_results"][model_name].get("vocab_compatible", False):
                result = self.train_single_model(model_name, training_dataset)
                training_results[model_name] = result
                
                if result["success"]:
                    successful_models.append(model_name)
        
        print(f"\n📊 Training Summary:")
        print(f"✅ Successfully trained: {len(successful_models)}/{len(training_results)}")
        print(f"✅ Models: {', '.join(successful_models)}")
        
        return {
            "individual_results": training_results,
            "successful_models": successful_models,
            "success_rate": len(successful_models) / len(training_results) if training_results else 0
        }

# Initialize production trainer
print("🚀 Production Model Training (Document 13)")
print("=" * 50)

# Only proceed if we have compatible models
if MODEL_COMPATIBILITY_RESULTS["overall_compatible"]:
    # Initialize trainer
    production_trainer = ProductionModelTrainer(
        compatibility_results=MODEL_COMPATIBILITY_RESULTS,
        kg_enhancer=KG_ENHANCER
    )
    
    # Prepare training data
    training_dataset = production_trainer.prepare_training_data(DATASET_RESULTS)
    
    # Train all compatible models
    TRAINING_RESULTS = production_trainer.train_all_compatible_models(training_dataset)
    
    print(f"\n🎯 Training Completed Successfully")
    print(f"📊 Success Rate: {TRAINING_RESULTS['success_rate']:.2%}")
    
else:
    print("❌ Cannot proceed with training - model compatibility issues detected")
    print("💡 Please resolve compatibility issues before training")
    TRAINING_RESULTS = {"success": False, "error": "Compatibility issues"}

print("=" * 50)
````

````python
# 4. Enhanced Dataset Preparation & Advanced Model Management (Phase 0B)

def check_github_model_availability(repo_url: str, version: str, model_type: str) -> Dict[str, Any]:
    """
    Check if models are available on GitHub releases.
    
    Parameters
    ----------
    repo_url : str
        GitHub repository URL for models
    version : str
        Model version to check
    model_type : str
        Type of model ("tokenizers" or "finetuned")
        
    Returns
    -------
    Dict[str, Any]
        Availability status and download information
    """
    if not repo_url or not MODEL_CONFIG["github_token"]:
        return {"available": False, "reason": "No GitHub configuration"}
    
    try:
        # Extract repo info from URL
        # Expected format: https://github.com/owner/repo or similar
        parts = repo_url.rstrip('/').split('/')
        if len(parts) < 2:
            return {"available": False, "reason": "Invalid repository URL"}
        
        owner, repo = parts[-2], parts[-1]
        
        # Check for release with version tag
        releases_url = f"https://api.github.com/repos/{owner}/{repo}/releases"
        headers = {"Authorization": f"token {MODEL_CONFIG['github_token']}"}
        
        response = requests.get(releases_url, headers=headers)
        if response.status_code != 200:
            return {"available": False, "reason": f"API error: {response.status_code}"}
        
        releases = response.json()
        target_tag = f"v{version}"
        
        for release in releases:
            if release["tag_name"] == target_tag:
                # Look for model zip file in assets
                expected_filename = f"{model_type}_v{version}.zip"
                for asset in release.get("assets", []):
                    if asset["name"] == expected_filename:
                        return {
                            "available": True,
                            "download_url": asset["browser_download_url"],
                            "filename": expected_filename,
                            "release_id": release["id"]
                        }
        
        return {"available": False, "reason": f"Version {version} not found in releases"}
        
    except Exception as e:
        return {"available": False, "reason": f"Error checking GitHub: {e}"}

def download_models_from_github(download_info: Dict[str, Any], extract_path: Path) -> bool:
    """
    Download and extract models from GitHub releases.
    
    Parameters
    ----------
    download_info : Dict[str, Any]
        Download information from check_github_model_availability
    extract_path : Path
        Path where to extract the models
        
    Returns
    -------
    bool
        True if download and extraction successful
    """
    try:
        print(f"[INFO] Downloading {download_info['filename']} from GitHub...")
        
        response = requests.get(download_info["download_url"], stream=True)
        response.raise_for_status()
        
        zip_path = extract_path.parent / download_info["filename"]
        
        with open(zip_path, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
        
        print(f"[INFO] Extracting {download_info['filename']}...")
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_path.parent)
        
        # Clean up zip file
        zip_path.unlink()
        
        print(f"[SUCCESS] Models downloaded and extracted successfully")
        return True
        
    except Exception as e:
        print(f"[ERROR] Failed to download models: {e}")
        return False

def check_model_version_compatibility(model_path: Path, tokenizer_path: Path, expected_version: str) -> Dict[str, Any]:
    """
    Enhanced model version compatibility checking.
    
    Parameters
    ----------
    model_path : Path
        Path to the fine-tuned model directory
    tokenizer_path : Path  
        Path to the tokenizer directory
    expected_version : str
        Expected version string (e.g., "1.0")
        
    Returns
    -------
    Dict[str, Any]
        Detailed compatibility information
    """
    result = {
        "compatible": False,
        "model_exists": model_path.exists(),
        "tokenizer_exists": tokenizer_path.exists(),
        "model_complete": False,
        "tokenizer_complete": False,
        "version_match": False,
        "paired": False,
        "issues": []
    }
    
    # Check if both paths exist
    if not result["model_exists"]:
        result["issues"].append(f"Model directory missing: {model_path}")
    if not result["tokenizer_exists"]:
        result["issues"].append(f"Tokenizer directory missing: {tokenizer_path}")
    
    if not (result["model_exists"] and result["tokenizer_exists"]):
        return result
    
    # Check for required model files (Phase 0B: Enhanced validation)
    required_model_files = ["config.json", "pytorch_model.bin", "training_args.bin"]
    optional_model_files = ["tokenizer_config.json"]  # Some models include this
    
    model_files_present = []
    for file in required_model_files:
        if (model_path / file).exists():
            model_files_present.append(file)
    
    result["model_complete"] = len(model_files_present) >= 2  # At least config and model
    
    # Check for required tokenizer files
    required_tokenizer_files = ["tokenizer_config.json", "vocab.txt"]
    optional_tokenizer_files = ["special_tokens_map.json", "tokenizer.json"]
    
    tokenizer_files_present = []
    for file in required_tokenizer_files + optional_tokenizer_files:
        if (tokenizer_path / file).exists():
            tokenizer_files_present.append(file)
    
    result["tokenizer_complete"] = len(tokenizer_files_present) >= 1
    
    # Check version compatibility
    version_indicators = [
        model_path / "version.txt",
        tokenizer_path / "version.txt",
        model_path / "config.json"
    ]
    
    for indicator in version_indicators:
        if indicator.exists():
            try:
                if indicator.name == "version.txt":
                    with open(indicator) as f:
                        version = f.read().strip()
                    if version == expected_version:
                        result["version_match"] = True
                        break
                elif indicator.name == "config.json":
                    with open(indicator) as f:
                        config = json.load(f)
                    # Check for version in config
                    if config.get("version") == expected_version:
                        result["version_match"] = True
                        break
            except Exception as e:
                result["issues"].append(f"Error reading version from {indicator}: {e}")
    
    # Phase 0B: Check if tokenizer and model are properly paired
    if result["model_complete"] and result["tokenizer_complete"]:
        try:
            # Quick compatibility test
            tokenizer = AutoTokenizer.from_pretrained(str(tokenizer_path))
            model = AutoModelForTokenClassification.from_pretrained(str(model_path))
            
            # Test tokenization
            test_text = "The polymer shows excellent properties."
            inputs = tokenizer(test_text, return_tensors="pt")
            outputs = model(**inputs)
            
            result["paired"] = True
            
            # Clean up
            del tokenizer, model, inputs, outputs
            torch.cuda.empty_cache() if torch.cuda.is_available() else None
            
        except Exception as e:
            result["issues"].append(f"Pairing test failed: {e}")
    
    # Overall compatibility
    result["compatible"] = (
        result["model_complete"] and 
        result["tokenizer_complete"] and 
        result["paired"] and
        (result["version_match"] or not MODEL_CONFIG["version_strict"])
    )
    
    return result

def get_real_datasets(override: bool = False, hours: int = 48) -> List[pd.DataFrame]:
    """
    Enhanced real dataset loading with better error handling.
    
    Parameters
    ----------
    override : bool
        If True, load all datasets regardless of modification time
    hours : int
        Only load datasets modified within this many hours (if override=False)
        
    Returns
    -------
    List[pd.DataFrame]
        List of loaded dataframes
    """
    now = datetime.now()
    real_data = []
    
    for folder in [TRAINING_DIR, TESTING_DIR]:
        if not folder.exists():
            print(f"[WARNING] Dataset folder not found: {folder}")
            continue
            
        csv_files = list(folder.glob("*.csv"))
        if not csv_files:
            print(f"[INFO] No CSV files found in: {folder}")
            continue
            
        for file in csv_files:
            try:
                mtime = datetime.fromtimestamp(file.stat().st_mtime)
                if override or (now - mtime) < timedelta(hours=hours):
                    df = pd.read_csv(file)
                    
                    # Enhanced validation
                    if "sentence" not in df.columns:
                        print(f"[WARNING] Skipping {file.name}: missing 'sentence' column")
                        continue
                    
                    # Clean and validate data
                    original_len = len(df)
                    df = df.dropna(subset=["sentence"]).reset_index(drop=True)
                    df = df[df["sentence"].str.len() > 10]  # Filter very short sentences
                    
                    if len(df) > 0:
                        real_data.append(df)
                        print(f"[INFO] Loaded dataset: {file.name} ({len(df)}/{original_len} valid samples)")
                    else:
                        print(f"[WARNING] No valid samples in {file.name}")
                        
            except Exception as e:
                print(f"[WARNING] Error loading {file.name}: {e}")
    
    return real_data

def generate_enhanced_synthetic_data(n: int = 25000) -> List[Dict[str, str]]:
    """
    Enhanced synthetic data generation with Phase 0B improvements.
    NOTE: MATERIAL entities completely removed.
    
    Parameters
    ----------
    n : int
        Number of synthetic samples to generate
        
    Returns
    -------
    List[Dict[str, str]]
        List of enhanced synthetic training samples
    """
    data = []
    
    # Phase 0B: Enhanced sentence patterns with better scientific context
    enhanced_templates = SENTENCE_TEMPLATES + [
        # Additional scientific templates for better training
        "The {polymer} exhibits a {property} of {value} {unit} when measured at standard conditions.",
        "Studies show that {polymer} has improved {property} reaching {value} {unit} under specific conditions.",
        "The measured {property} for {polymer} was {value} {unit}, indicating excellent performance.",
        "Research demonstrates {polymer} achieving {property} values of {value} {unit} consistently.",
        "Analysis revealed {polymer} with {property} = {value} {unit} showing superior characteristics.",
        "Experimental data shows {polymer} exhibiting {property} of approximately {value} {unit}.",
        "The {polymer} sample demonstrated {property} values ranging near {value} {unit}.",
        "Characterization of {polymer} revealed {property} measurements of {value} {unit}."
    ]
    
    # Phase 0B: Enhanced value formats with more scientific notation
    enhanced_values = VALUE_FORMATS + [
        "1.25 × 10^3", "2.5 × 10^-2", "7.8 × 10^6", "3.14 × 10^-4",
        "~5.0", "≈10.2", "±2.5", "5.5±0.3", "12.7 ± 1.2",
        "0.95", "1.05", "2.15", "3.75", "8.90", "15.6", "22.3"
    ]
    
    for i in range(n):
        # Select random components (MATERIAL completely removed)
        polymer = random.choice(POLYMER_NAMES)
        property_name = random.choice(PROPERTY_NAMES) 
        unit = random.choice(SCIENTIFIC_UNITS)
        symbol = random.choice(SCIENTIFIC_SYMBOLS)
        value = random.choice(enhanced_values)
        
        # Phase 0B: Use enhanced templates for better training
        template = random.choice(enhanced_templates)
        
        # Generate sentence using template
        try:
            sentence = template.format(
                polymer=polymer,
                property=property_name,
                unit=unit,
                symbol=symbol,
                value=value
            )
            
            data.append({
                "sentence": sentence,
                "polymer": polymer,
                "property": property_name,
                "unit": unit,
                "symbol": symbol,
                "value": value,
                # "material": material  # Completely removed
                "template_id": i % len(enhanced_templates),  # For debugging
                "enhanced": True  # Mark as Phase 0B enhanced sample
            })
            
        except KeyError as e:
            # Template might not have all placeholders, skip this combination
            continue
    
    print(f"[INFO] Generated {len(data)} enhanced synthetic samples (Phase 0B)")
    return data

print("[INFO] Enhanced dataset preparation functions loaded (Phase 0B)")
print("[INFO] GitHub model management functions loaded")
print("[INFO] MATERIAL entity support completely removed from synthetic data generation")
print("[INFO] Enhanced sentence templates and value formats added")
print("[INFO] Advanced model compatibility checking enabled")
````

````python
# 5. Enhanced Tokenization & Advanced Vocabulary Extension (Phase 0B)

def enhance_tokenizer_vocabulary(tokenizer, model_name: str) -> Any:
    """
    Enhanced tokenizer vocabulary extension using centralized constants.
    
    Parameters
    ----------
    tokenizer : AutoTokenizer
        Base tokenizer to enhance
    model_name : str
        Name of the model for logging
        
    Returns
    -------
    AutoTokenizer
        Enhanced tokenizer with extended vocabulary
    """
    print(f"[INFO] Enhancing tokenizer vocabulary for {model_name} using centralized constants...")
    
    # Use centralized scientific vocabulary
    scientific_vocab = set()
    
    # Add scientific units from centralized constants
    scientific_vocab.update(SCIENTIFIC_UNITS)
    
    # Add scientific symbols from centralized constants  
    scientific_vocab.update(SCIENTIFIC_SYMBOLS)
    
    # Add polymer names from centralized constants
    scientific_vocab.update(POLYMER_NAMES[:500])  # Use first 500 to avoid bloat
    
    # Add property names from centralized constants
    scientific_vocab.update(PROPERTY_NAMES)
    
    # Add scientific measurement terms from centralized patterns
    measurement_terms = set()
    for pattern_name, units in SEMANTIC_COHERENCE_PATTERNS["value_unit_pairs"].items():
        measurement_terms.update(units)
    scientific_vocab.update(measurement_terms)
    
    # Add scientific transition words from centralized stopwords
    scientific_vocab.update(SCIENTIFIC_STOPWORDS["scientific_transitions"])
    scientific_vocab.update(SCIENTIFIC_STOPWORDS["measurement_contexts"])
    
    # Add special scientific characters from sentence splitting enhancements
    scientific_vocab.update(SENTENCE_SPLITTING_ENHANCEMENTS["special_characters"]["greek_letters"])
    scientific_vocab.update(SENTENCE_SPLITTING_ENHANCEMENTS["special_characters"]["mathematical"])
    scientific_vocab.update(SENTENCE_SPLITTING_ENHANCEMENTS["special_characters"]["chemical"])
    
    # Filter existing vocabulary to avoid duplicates
    existing_vocab = set(tokenizer.get_vocab().keys())
    new_tokens = scientific_vocab - existing_vocab
    
    # Remove tokens that are too short or contain problematic characters
    filtered_tokens = []
    for token in new_tokens:
        if (len(token) >= 2 and 
            not token.startswith(('##', '[', '<')) and
            not token.endswith((']', '>')) and
            len(token) <= 25):  # Reasonable length limit
            filtered_tokens.append(token)
    
    if filtered_tokens:
        print(f"[INFO] Adding {len(filtered_tokens)} new scientific terms to vocabulary")
        tokenizer.add_tokens(filtered_tokens)
        print(f"[INFO] Tokenizer vocabulary expanded from {len(existing_vocab)} to {len(tokenizer)} tokens")
    else:
        print(f"[INFO] No new tokens needed - vocabulary already comprehensive")
    
    return tokenizer

def enhanced_tokenize_and_label(sample: Dict[str, Any], tokenizer, model_name: str) -> Dict[str, Any]:
    """
    Enhanced tokenization with advanced pattern recognition using centralized constants.
    
    Parameters
    ----------
    sample : Dict[str, Any]
        Sample containing text and entities
    tokenizer : AutoTokenizer
        Enhanced tokenizer
    model_name : str
        Model name for logging
        
    Returns
    -------
    Dict[str, Any]
        Tokenized sample with enhanced labels
    """
    text = sample.get("text", "")
    entities = sample.get("entities", [])
    
    # Enhanced preprocessing using centralized patterns
    enhanced_text = text
    
    # Apply scientific notation normalization using centralized patterns
    for pattern_name, pattern in TOKENIZATION_PATTERNS.items():
        if pattern_name == "scientific_notation":
            # Normalize scientific notation format
            enhanced_text = re.sub(pattern, lambda m: m.group(0).replace(" ", ""), enhanced_text)
    
    # Preserve scientific abbreviations using centralized exceptions
    for abbrev in SENTENCE_SPLITTING_ENHANCEMENTS["abbreviation_exceptions"]:
        # Ensure abbreviations are not split inappropriately
        enhanced_text = enhanced_text.replace(f"{abbrev} ", f"{abbrev}__SPACE__")
    
    # Tokenize with enhanced text
    tokenized = tokenizer(
        enhanced_text,
        truncation=True,
        padding=False,
        max_length=512,
        return_offsets_mapping=True,
        add_special_tokens=True
    )
    
    # Restore spaces in abbreviations
    enhanced_text = enhanced_text.replace("__SPACE__", " ")
    
    tokens = tokenizer.convert_ids_to_tokens(tokenized["input_ids"])
    labels = ["O"] * len(tokens)
    
    # Enhanced entity labeling with semantic coherence validation
    for entity in entities:
        entity_text = entity.get("text", "")
        entity_label = entity.get("label", "")
        entity_start = entity.get("start", 0)
        entity_end = entity.get("end", 0)
        
        # Skip MATERIAL entities (Phase 0 compliance)
        if entity_label == "MATERIAL":
            continue
        
        # Apply enhanced confidence thresholds from centralized constants
        confidence = entity.get("confidence", 1.0)
        threshold_config = ENHANCED_THRESHOLDS.get(entity_label, {})
        base_threshold = threshold_config.get("base_threshold", 0.75)
        
        # Apply semantic boost for coherent contexts
        semantic_boost = 0.0
        if entity_label in ["VALUE", "UNIT"] and threshold_config:
            # Check for VALUE-UNIT semantic coherence
            for pattern_name, units in SEMANTIC_COHERENCE_PATTERNS["value_unit_pairs"].items():
                if any(unit in entity_text.lower() for unit in units):
                    semantic_boost = threshold_config.get("semantic_boost", 0.0)
                    break
        
        adjusted_confidence = confidence + semantic_boost
        
        # Apply confidence threshold
        if adjusted_confidence < base_threshold:
            continue
        
        # Apply over-merging prevention using centralized criteria
        entity_length = len(entity_text)
        max_length = OVER_MERGING_PREVENTION.get("max_entity_length", 50)
        if entity_length > max_length:
            continue
        
        # Check for forbidden word combinations
        entity_words = entity_text.split()
        max_words = OVER_MERGING_PREVENTION.get("max_word_count", 8)
        if len(entity_words) > max_words:
            continue
        
        # Map entity spans to token positions with enhanced handling
        offset_mapping = tokenized["offset_mapping"]
        token_start = None
        token_end = None
        
        for i, (start, end) in enumerate(offset_mapping):
            if start <= entity_start < end and token_start is None:
                token_start = i
            if start < entity_end <= end:
                token_end = i
                break
        
        if token_start is not None and token_end is not None:
            # Apply BIO labeling scheme with enhanced validation
            for j in range(token_start, token_end + 1):
                if j < len(labels):
                    if j == token_start:
                        labels[j] = f"B-{entity_label}"
                    else:
                        labels[j] = f"I-{entity_label}"
    
    # Convert labels to IDs
    label_ids = [LABEL2ID.get(label, LABEL2ID["O"]) for label in labels]
    
    return {
        "input_ids": tokenized["input_ids"],
        "attention_mask": tokenized["attention_mask"],
        "labels": label_ids,
        "tokens": tokens,
        "enhanced": True,
        "model_name": model_name
    }

def to_enhanced_dataset(encoded_samples: List[Dict[str, Any]]) -> Dataset:
    """
    Convert enhanced encoded samples to HuggingFace Dataset.
    
    Parameters
    ----------
    encoded_samples : List[Dict[str, Any]]
        List of enhanced encoded samples
        
    Returns
    -------
    Dataset
        HuggingFace Dataset ready for training
    """
    # Flatten the data for Dataset creation
    input_ids = [sample["input_ids"] for sample in encoded_samples]
    attention_masks = [sample["attention_mask"] for sample in encoded_samples] 
    labels = [sample["labels"] for sample in encoded_samples]
    
    return Dataset.from_dict({
        "input_ids": input_ids,
        "attention_mask": attention_masks,
        "labels": labels
    })

def enhanced_scientific_text_processing(text: str) -> str:
    """
    Enhanced scientific text processing using centralized patterns.
    
    Parameters
    ----------
    text : str
        Raw scientific text
        
    Returns
    -------
    str
        Processed text with enhanced scientific formatting
    """
    processed_text = text
    
    # Apply tokenization patterns for scientific notation
    for pattern_name, pattern in TOKENIZATION_PATTERNS.items():
        if "scientific" in pattern_name or "measurement" in pattern_name:
            # Normalize spacing around scientific notation
            processed_text = re.sub(
                pattern, 
                lambda m: m.group(0).strip(),
                processed_text
            )
    
    # Preserve chemical formulas and polymer names using context preservation
    for context_name, pattern in SENTENCE_SPLITTING_ENHANCEMENTS["context_preservation"].items():
        # Ensure scientific contexts are preserved
        matches = re.findall(pattern, processed_text)
        for match in matches:
            if len(match) > 2:  # Skip very short matches
                # Preserve the match by replacing spaces with underscores temporarily
                placeholder = match.replace(" ", "_PRESERVE_")
                processed_text = processed_text.replace(match, placeholder)
    
    # Apply scientific stopword handling
    words = processed_text.split()
    processed_words = []
    
    for word in words:
        # Restore preserved spaces
        if "_PRESERVE_" in word:
            word = word.replace("_PRESERVE_", " ")
        
        # Handle scientific stopwords appropriately
        word_lower = word.lower()
        if word_lower in SCIENTIFIC_STOPWORDS["scientific_preserve"]:
            # Preserve scientific units and measurements as-is
            processed_words.append(word)
        elif word_lower in SCIENTIFIC_STOPWORDS["scientific_transitions"]:
            # Preserve scientific transition words
            processed_words.append(word)
        elif word_lower in SCIENTIFIC_STOPWORDS["measurement_contexts"]:
            # Preserve measurement context words
            processed_words.append(word)
        else:
            # Regular word processing
            processed_words.append(word)
    
    return " ".join(processed_words)

# Display enhanced tokenization configuration
print(f"[ENHANCED TOKENIZATION - Phase 0B with Centralized Constants]")
print(f"✅ Scientific vocabulary sources: {len(SCIENTIFIC_UNITS)} units, {len(SCIENTIFIC_SYMBOLS)} symbols")
print(f"✅ Polymer names: {len(POLYMER_NAMES)} available")
print(f"✅ Property names: {len(PROPERTY_NAMES)} available")
print(f"✅ Tokenization patterns: {len(TOKENIZATION_PATTERNS)} patterns")
print(f"✅ Semantic coherence categories: {len(SEMANTIC_COHERENCE_PATTERNS['value_unit_pairs'])} categories")
print(f"✅ Scientific stopword categories: {len(SCIENTIFIC_STOPWORDS)} categories")
print(f"✅ Sentence splitting enhancements: {len(SENTENCE_SPLITTING_ENHANCEMENTS)} improvements")
print(f"✅ Enhanced thresholds: {len(ENHANCED_THRESHOLDS)} entity types")
print(f"✅ Over-merging prevention: {len(OVER_MERGING_PREVENTION)} criteria")

print(f"\n[SUCCESS] Enhanced tokenization functions loaded with centralized constants")
````

````python
# 6. Enhanced Training Pipeline with Advanced Monitoring (Phase 0B)

def check_existing_enhanced_models() -> Dict[str, Dict[str, Any]]:
    """
    Enhanced model checking with detailed compatibility analysis using centralized constants.
    
    Returns
    -------
    Dict[str, Dict[str, Any]]
        Detailed status for each model including compatibility information
    """
    model_status = {}
    
    print("[INFO] Checking existing models with enhanced validation using centralized constants...")
    
    for model_cfg in ENSEMBLE_MODELS:
        model_path = OUTPUT_DIR / model_cfg.name
        tokenizer_path = TOKENIZERS_DIR / f"{model_cfg.name}_extended"
        
        compatibility_info = check_model_version_compatibility(
            model_path, 
            tokenizer_path, 
            MODEL_CONFIG["models_version"]
        )
        
        model_status[model_cfg.name] = {
            "config": model_cfg,
            "compatibility": compatibility_info,
            "needs_training": not compatibility_info["compatible"],
            "needs_retraining": MODEL_CONFIG["force_retrain"]
        }
        
        # Enhanced status reporting
        if compatibility_info["compatible"]:
            print(f"✅ {model_cfg.name}: Available and compatible")
            if compatibility_info["version_match"]:
                print(f"   └── Version: {MODEL_CONFIG['models_version']} ✅")
            if compatibility_info["paired"]:
                print(f"   └── Tokenizer paired: ✅")
        else:
            print(f"❌ {model_cfg.name}: Issues detected")
            for issue in compatibility_info["issues"]:
                print(f"   └── {issue}")
            
            # Check if available on GitHub
            if MODEL_CONFIG["upload_to_github"]:
                github_status = check_github_model_availability(
                    MODEL_CONFIG["finetuned_remote_url"],
                    MODEL_CONFIG["models_version"],
                    "finetuned"
                )
                if github_status["available"]:
                    print(f"   └── Available on GitHub: ✅")
                else:
                    print(f"   └── GitHub status: {github_status['reason']}")
    
    return model_status

def train_enhanced_model(model_cfg, train_samples: List[Dict], val_samples: List[Dict]) -> Dict[str, Any]:
    """
    Enhanced model training with Phase 0B improvements using centralized training configuration.
    
    Parameters
    ----------
    model_cfg : ModelConfig
        Configuration for the model to train
    train_samples : List[Dict]
        Enhanced training samples
    val_samples : List[Dict] 
        Enhanced validation samples
        
    Returns
    -------
    Dict[str, Any]
        Training results and statistics
    """
    training_result = {
        "success": False,
        "model_name": model_cfg.name,
        "training_time": 0,
        "final_eval_loss": None,
        "best_eval_loss": None,
        "total_epochs": 0,
        "total_steps": 0,
        "enhanced_features": []
    }
    
    start_time = datetime.now()
    
    try:
        print(f"\n[TRAINING] Enhanced training for {model_cfg.name} ({model_cfg.model_id})")
        print(f"[INFO] Phase 0B features enabled with centralized configuration")
        
        # Load and enhance tokenizer
        print(f"[INFO] Loading and enhancing tokenizer...")
        tokenizer = AutoTokenizer.from_pretrained(model_cfg.model_id, use_fast=True)
        
        # Ensure tokenizer has pad token
        if tokenizer.pad_token is None:
            tokenizer.pad_token = tokenizer.eos_token
            training_result["enhanced_features"].append("pad_token_added")
        
        # Phase 0B: Enhanced vocabulary using centralized constants
        if MODEL_CONFIG["improved_vocabulary_extension"]:
            tokenizer = enhance_tokenizer_vocabulary(tokenizer, model_cfg.name)
            training_result["enhanced_features"].append("vocabulary_enhanced")
        
        # Load model with enhanced configuration
        print(f"[INFO] Loading model with enhanced configuration...")
        model = AutoModelForTokenClassification.from_pretrained(
            model_cfg.model_id,
            num_labels=len(LABELS),
            id2label=ID2LABEL,
            label2id=LABEL2ID,
            ignore_mismatched_sizes=True
        )
        
        # Phase 0B: Resize model embeddings if vocabulary was extended
        if len(tokenizer) > model.config.vocab_size:
            print(f"[INFO] Resizing model embeddings: {model.config.vocab_size} -> {len(tokenizer)}")
            model.resize_token_embeddings(len(tokenizer))
            training_result["enhanced_features"].append("embeddings_resized")
        
        print(f"[INFO] Enhanced tokenizing datasets using centralized patterns...")
        
        # Enhanced tokenization with Phase 0B improvements using centralized functions
        train_encoded = [
            enhanced_tokenize_and_label(sample, tokenizer, model_cfg.name) 
            for sample in train_samples
        ]
        val_encoded = [
            enhanced_tokenize_and_label(sample, tokenizer, model_cfg.name) 
            for sample in val_samples
        ]
        
        train_dataset = to_enhanced_dataset(train_encoded)
        val_dataset = to_enhanced_dataset(val_encoded)
        
        print(f"[INFO] Enhanced dataset sizes - Train: {len(train_dataset)}, Val: {len(val_dataset)}")
        
        # Setup enhanced output directory
        model_output_path = OUTPUT_DIR / model_cfg.name
        model_output_path.mkdir(parents=True, exist_ok=True)
        
        # Phase 0B: Enhanced training arguments using centralized configuration
        training_config = model_cfg.training_config
        
        # Calculate dynamic learning rate using centralized thresholds
        base_lr = training_config.get("lr", MODEL_TRAINING_CONFIG["default_learning_rate"])
        dataset_size = len(train_dataset)
        
        large_threshold = MODEL_TRAINING_CONFIG["dataset_size_adjustments"]["large_dataset_threshold"]
        small_threshold = MODEL_TRAINING_CONFIG["dataset_size_adjustments"]["small_dataset_threshold"]
        
        if dataset_size > large_threshold:
            lr_factor = MODEL_TRAINING_CONFIG["dataset_size_adjustments"]["large_dataset_lr_factor"]
            adjusted_lr = base_lr * lr_factor
        elif dataset_size < small_threshold:
            lr_factor = MODEL_TRAINING_CONFIG["dataset_size_adjustments"]["small_dataset_lr_factor"]
            adjusted_lr = base_lr * lr_factor
        else:
            adjusted_lr = base_lr
        
        # Enhanced training arguments using centralized configuration
        training_args_config = MODEL_TRAINING_CONFIG
        
        args = TrainingArguments(
            output_dir=str(model_output_path),
            eval_strategy=training_args_config["model_saving"]["eval_strategy"],
            save_strategy=training_args_config["model_saving"]["save_strategy"],
            learning_rate=adjusted_lr,
            per_device_train_batch_size=training_config.get("batch_size", training_args_config["default_batch_size"]),
            per_device_eval_batch_size=training_config.get("eval_batch_size", training_args_config["default_eval_batch_size"]),
            num_train_epochs=training_config.get("epochs", training_args_config["default_epochs"]),
            weight_decay=training_config.get("weight_decay", training_args_config["default_weight_decay"]),
            gradient_accumulation_steps=training_config.get("grad_accum_steps", training_args_config["default_gradient_accumulation_steps"]),
            logging_dir=str(model_output_path / "logs"),
            logging_steps=training_args_config["advanced_features"]["logging_steps"],
            logging_first_step=training_args_config["advanced_features"]["logging_first_step"],
            load_best_model_at_end=training_args_config["model_saving"]["load_best_model_at_end"],
            metric_for_best_model=training_args_config["model_saving"]["metric_for_best_model"],
            greater_is_better=training_args_config["model_saving"]["greater_is_better"],
            save_total_limit=training_args_config["model_saving"]["save_total_limit"],
            fp16=torch.cuda.is_available() and training_args_config["hardware_optimization"]["fp16"],
            dataloader_pin_memory=training_args_config["hardware_optimization"]["dataloader_pin_memory"],
            dataloader_num_workers=training_args_config["hardware_optimization"]["dataloader_num_workers"],
            remove_unused_columns=training_args_config["hardware_optimization"]["remove_unused_columns"],
            # Phase 0B: Enhanced training settings using centralized configuration
            warmup_ratio=training_args_config["advanced_features"]["warmup_ratio"],
            lr_scheduler_type=training_args_config["advanced_features"]["lr_scheduler_type"],
            label_smoothing_factor=training_args_config["advanced_features"]["label_smoothing_factor"],
            report_to=["wandb"] if MODEL_CONFIG.get("wandb_enabled", False) else [],
            run_name=f"{model_cfg.name}_enhanced_v{MODEL_CONFIG['models_version']}"
        )
        
        # Phase 0B: Enhanced data collator
        data_collator = DataCollatorForTokenClassification(
            tokenizer, 
            pad_to_multiple_of=8,
            return_tensors="pt"
        )
        
        # Enhanced callbacks using centralized configuration
        early_stopping_config = MODEL_TRAINING_CONFIG["early_stopping"]
        callbacks = [
            EarlyStoppingCallback(
                early_stopping_patience=early_stopping_config["patience"], 
                early_stopping_threshold=early_stopping_config["threshold"]
            )
        ]
        
        # Initialize enhanced trainer
        trainer = Trainer(
            model=model,
            args=args,
            train_dataset=train_dataset,
            eval_dataset=val_dataset,
            tokenizer=tokenizer,
            data_collator=data_collator,
            callbacks=callbacks
        )
        
        print(f"[INFO] Starting enhanced training with centralized configuration...")
        print(f"[INFO] Learning rate: {adjusted_lr} (adjusted from {base_lr})")
        print(f"[INFO] Dataset size: {dataset_size} samples")
        
        # Train model with enhanced monitoring
        training_output = trainer.train()
        
        # Extract training statistics
        training_result["total_epochs"] = args.num_train_epochs
        training_result["total_steps"] = training_output.global_step
        training_result["final_eval_loss"] = training_output.training_loss
        
        # Get best evaluation loss from logs
        if trainer.state.log_history:
            eval_losses = [log.get("eval_loss") for log in trainer.state.log_history if "eval_loss" in log]
            if eval_losses:
                training_result["best_eval_loss"] = min(eval_losses)
        
        print(f"[INFO] Enhanced training completed - saving model and tokenizer...")
        
        # Save model with enhanced metadata
        trainer.save_model(str(model_output_path))
        
        # Save enhanced tokenizer
        tokenizer_output_path = TOKENIZERS_DIR / f"{model_cfg.name}_extended"
        tokenizer_output_path.mkdir(parents=True, exist_ok=True)
        tokenizer.save_pretrained(str(tokenizer_output_path))
        
        # Phase 0B: Save enhanced metadata using centralized constants
        metadata = {
            "model_name": model_cfg.name,
            "model_id": model_cfg.model_id,
            "version": MODEL_CONFIG["models_version"],
            "training_date": datetime.now().isoformat(),
            "enhanced_features": training_result["enhanced_features"],
            "dataset_stats": {
                "train_samples": len(train_dataset),
                "val_samples": len(val_dataset),
                "total_samples": len(train_dataset) + len(val_dataset)
            },
            "training_config": training_config,
            "centralized_config": {
                "enhanced_thresholds": len(ENHANCED_THRESHOLDS),
                "scientific_stopwords": len(SCIENTIFIC_STOPWORDS),
                "tokenization_patterns": len(TOKENIZATION_PATTERNS),
                "semantic_coherence": len(SEMANTIC_COHERENCE_PATTERNS),
                "phase_0b_features": MODEL_DEFAULT_CONFIG["phase_0b_features"]
            },
            "labels": LABELS,
            "material_removed": True,  # Mark Phase 0 compliance
            "phase_0b_enhanced": True,
            "centralized_constants": True  # Mark use of centralized constants
        }
        
        # Save metadata files
        with open(model_output_path / "metadata.json", 'w') as f:
            json.dump(metadata, f, indent=2)
        
        with open(model_output_path / "version.txt", 'w') as f:
            f.write(MODEL_CONFIG["models_version"])
            
        with open(tokenizer_output_path / "version.txt", 'w') as f:
            f.write(MODEL_CONFIG["models_version"])
        
        training_result["training_time"] = (datetime.now() - start_time).total_seconds()
        training_result["success"] = True
        
        print(f"[SUCCESS] {model_cfg.name} enhanced training completed with centralized configuration")
        print(f"  - Model saved to: {model_output_path}")
        print(f"  - Tokenizer saved to: {tokenizer_output_path}")
        print(f"  - Training time: {training_result['training_time']:.1f}s")
        print(f"  - Enhanced features: {', '.join(training_result['enhanced_features'])}")
        print(f"  - Centralized constants: ✅")
        
        # Memory cleanup
        del trainer, model, tokenizer, train_dataset, val_dataset
        torch.cuda.empty_cache() if torch.cuda.is_available() else None
        gc.collect()
        
        return training_result
        
    except Exception as e:
        training_result["training_time"] = (datetime.now() - start_time).total_seconds()
        training_result["error"] = str(e)
        print(f"[ERROR] Enhanced training failed for {model_cfg.name}: {e}")
        import traceback
        traceback.print_exc()
        return training_result

# Main enhanced training execution using centralized configuration
print("[INFO] Starting enhanced model training pipeline (Phase 0B with centralized constants)...")

# Check existing models with enhanced validation
existing_models = check_existing_enhanced_models()
models_to_train = [
    model_status["config"] for model_name, model_status in existing_models.items()
    if model_status["needs_training"] or model_status["needs_retraining"]
]

if not models_to_train:
    print("[INFO] All models are already available and compatible with Phase 0B standards.")
    print("[INFO] Set force_retrain=True to retrain with Phase 0B enhancements.")
else:
    print(f"[INFO] Training {len(models_to_train)} models with Phase 0B enhancements using centralized constants:")
    for model_cfg in models_to_train:
        print(f"  - {model_cfg.name} ({model_cfg.model_id})")
    
    # Prepare enhanced training data
    train_samples, val_samples = prepare_enhanced_training_data(
        force_synthetic=False,
        synthetic_count=25000
    )
    
    # Train each model with enhanced pipeline using centralized configuration
    training_results = []
    successful_trainings = 0
    failed_trainings = 0
    
    for model_cfg in models_to_train:
        print(f"\n{'='*60}")
        print(f"ENHANCED TRAINING: {model_cfg.name}")
        print(f"{'='*60}")
        
        result = train_enhanced_model(model_cfg, train_samples, val_samples)
        training_results.append(result)
        
        if result["success"]:
            successful_trainings += 1
            print(f"✅ {model_cfg.name}: Training successful")
        else:
            failed_trainings += 1
            print(f"❌ {model_cfg.name}: Training failed")
    
    print(f"\n[ENHANCED TRAINING SUMMARY - Phase 0B with Centralized Constants]")
    print(f"  - Successful: {successful_trainings}")
    print(f"  - Failed: {failed_trainings}") 
    print(f"  - Total: {len(models_to_train)}")
    print(f"  - Enhanced features applied: ✅")
    print(f"  - MATERIAL entities removed: ✅")
    print(f"  - Centralized constants: ✅")
    print(f"  - Version: {MODEL_CONFIG['models_version']}")
    
    # Save training summary with centralized configuration info
    summary = {
        "phase": "0B",
        "timestamp": datetime.now().isoformat(),
        "total_models": len(models_to_train),
        "successful": successful_trainings,
        "failed": failed_trainings,
        "version": MODEL_CONFIG["models_version"],
        "enhanced_features_enabled": True,
        "material_entities_removed": True,
        "centralized_constants_used": True,
        "configuration_sources": {
            "enhanced_thresholds": len(ENHANCED_THRESHOLDS),
            "scientific_stopwords": len(SCIENTIFIC_STOPWORDS), 
            "tokenization_patterns": len(TOKENIZATION_PATTERNS),
            "training_config": len(MODEL_TRAINING_CONFIG),
            "github_config": len(GITHUB_INTEGRATION_CONFIG)
        },
        "results": training_results
    }
    
    with open(EXPORTS_DIR / f"training_summary_v{MODEL_CONFIG['models_version']}.json", 'w') as f:
        json.dump(summary, f, indent=2)
    
    if failed_trainings == 0:
        print(f"\n[SUCCESS] All models trained successfully with Phase 0B enhancements and centralized constants!")
        print(f"[NEXT] Proceed to model export and GitHub integration")
    else:
        print(f"\n[WARNING] {failed_trainings} models failed enhanced training")
        print(f"[ACTION] Review training logs and retry failed models")
````

````python
# 7. Model Export & Packaging

def package_trained_models() -> bool:
    """
    Package all trained models and tokenizers into zip files for distribution.
    
    Returns
    -------
    bool
        True if packaging successful
    """
    try:
        print("[INFO] Packaging trained models...")
        
        # Create zip files
        finetuned_zip = WORKSPACE_DIR / f"finetuned_models_v{MODEL_CONFIG['models_version']}.zip"
        tokenizers_zip = WORKSPACE_DIR / f"tokenizers_v{MODEL_CONFIG['models_version']}.zip"
        
        # Package fine-tuned models
        with zipfile.ZipFile(finetuned_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
            for model_path in OUTPUT_DIR.iterdir():
                if model_path.is_dir():
                    for file_path in model_path.rglob('*'):
                        if file_path.is_file():
                            arcname = file_path.relative_to(OUTPUT_DIR)
                            zipf.write(file_path, arcname)
        
        # Package tokenizers
        with zipfile.ZipFile(tokenizers_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
            for tokenizer_path in TOKENIZERS_DIR.iterdir():
                if tokenizer_path.is_dir():
                    for file_path in tokenizer_path.rglob('*'):
                        if file_path.is_file():
                            arcname = file_path.relative_to(TOKENIZERS_DIR)
                            zipf.write(file_path, arcname)
        
        print(f"[SUCCESS] Models packaged successfully:")
        print(f"  - Fine-tuned models: {finetuned_zip}")
        print(f"  - Tokenizers: {tokenizers_zip}")
        
        # Copy to Google Drive if in Colab
        if 'google.colab' in sys.modules:
            try:
                drive_dir = Path('/content/drive/MyDrive/polymer_nlp_extractor')
                drive_dir.mkdir(parents=True, exist_ok=True)
                
                import shutil
                shutil.copy(finetuned_zip, drive_dir / finetuned_zip.name)
                shutil.copy(tokenizers_zip, drive_dir / tokenizers_zip.name)
                
                print(f"[INFO] Models also saved to Google Drive")
            except Exception as e:
                print(f"[WARNING] Failed to copy to Google Drive: {e}")
        
        return True
        
    except Exception as e:
        print(f"[ERROR] Model packaging failed: {e}")
        return False

def verify_trained_models() -> Dict[str, Dict[str, Any]]:
    """
    Verify all trained models are complete and functional.
    
    Returns
    -------
    Dict[str, Dict[str, Any]]
        Verification results for each model
    """
    verification_results = {}
    
    print("[INFO] Verifying trained models...")
    
    for model_cfg in ENSEMBLE_MODELS:
        model_name = model_cfg.name
        model_path = OUTPUT_DIR / model_name
        tokenizer_path = TOKENIZERS_DIR / f"{model_name}_extended"
        
        result = {
            "model_exists": model_path.exists(),
            "tokenizer_exists": tokenizer_path.exists(),
            "model_files_complete": False,
            "tokenizer_files_complete": False,
            "loadable": False,
            "version_match": False
        }
        
        # Check required files
        if result["model_exists"]:
            required_model_files = ["config.json", "pytorch_model.bin"]
            result["model_files_complete"] = all(
                (model_path / f).exists() for f in required_model_files
            )
        
        if result["tokenizer_exists"]:
            required_tokenizer_files = ["tokenizer_config.json"]
            result["tokenizer_files_complete"] = all(
                (tokenizer_path / f).exists() for f in required_tokenizer_files
            )
        
        # Test loading
        if result["model_files_complete"] and result["tokenizer_files_complete"]:
            try:
                from transformers import AutoTokenizer, AutoModelForTokenClassification
                
                tokenizer = AutoTokenizer.from_pretrained(str(tokenizer_path))
                model = AutoModelForTokenClassification.from_pretrained(str(model_path))
                
                # Quick inference test
                test_text = "The polymer shows excellent properties."
                inputs = tokenizer(test_text, return_tensors="pt")
                outputs = model(**inputs)
                
                result["loadable"] = True
                
                # Clean up
                del tokenizer, model, inputs, outputs
                torch.cuda.empty_cache() if torch.cuda.is_available() else None
                
            except Exception as e:
                print(f"[WARNING] {model_name} failed loading test: {e}")
        
        # Check version
        version_file = model_path / "version.txt"
        if version_file.exists():
            try:
                with open(version_file) as f:
                    model_version = f.read().strip()
                result["version_match"] = model_version == MODEL_CONFIG["models_version"]
            except Exception:
                pass
        
        verification_results[model_name] = result
        
        # Print status
        if all([result["model_files_complete"], result["tokenizer_files_complete"], result["loadable"]]):
            print(f"✅ {model_name}: Verified successfully")
        else:
            issues = []
            if not result["model_files_complete"]:
                issues.append("model files incomplete")
            if not result["tokenizer_files_complete"]:
                issues.append("tokenizer files incomplete")
            if not result["loadable"]:
                issues.append("loading failed")
            print(f"❌ {model_name}: Issues - {', '.join(issues)}")
    
    return verification_results

# Execute export and verification
print("[INFO] Starting model export and verification...")

# Verify models
verification_results = verify_trained_models()

# Count successful models
successful_models = sum(
    1 for result in verification_results.values()
    if result["loadable"] and result["model_files_complete"] and result["tokenizer_files_complete"]
)

print(f"\n[VERIFICATION SUMMARY]")
print(f"  - Verified models: {successful_models}/{len(ENSEMBLE_MODELS)}")
print(f"  - Models version: {MODEL_CONFIG['models_version']}")

if successful_models == len(ENSEMBLE_MODELS):
    print("[SUCCESS] All models verified successfully!")
    
    # Package models
    if package_trained_models():
        print("[SUCCESS] Training pipeline completed successfully!")
        print("\n[NEXT STEPS]")
        print("1. Models are ready for use in the inference pipeline")
        print("2. Run the main pipeline notebook for inference and evaluation")
        print("3. Models are packaged and saved for distribution")
    else:
        print("[WARNING] Model packaging failed")
else:
    print(f"[ERROR] {len(ENSEMBLE_MODELS) - successful_models} models failed verification")
    print("[ACTION] Review training logs and retry failed models")
````
